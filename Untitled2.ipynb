```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("queue_totals.csv")
df = df.reset_index(drop=False)

df.head()
```

```python
df.drop(columns=["timestamp", "dropped_high_in_low","low_in_high_queue"], inplace=True)
```

```python
df["total_high_in_system"]= df["high_in_high_queue"]+df["high_in_low_queue"]
```

```python
df.head(200)
```

```python
df.plot(
    x="index", 
    y=["total_high_in_system", "low_in_low_queue"], 
    kind="line"
)
plt.show()
```
