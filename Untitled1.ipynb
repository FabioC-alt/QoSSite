```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV file
df = pd.read_csv("load_test_results.csv")

# Calculate total requests per batch
df["Total_Requests"] = df["Status_200"] + df["Status_503"]

# Plot
plt.figure(figsize=(10, 6))
plt.plot(df["Total_Requests"], df["Status_200"], marker="o", color="blue", linewidth=2, label="200 OK")

# Annotate points with batch number
for i, row in df.iterrows():
    plt.text(row["Total_Requests"] + 2, row["Status_200"], f"B{row['Batch']}", va="center")

# Labels and title
plt.xlabel("Total Requests Sent (per batch)")
plt.ylabel("Successful Requests (200)")
plt.title("Successful Requests vs Total Requests")
plt.grid(True)
plt.legend()
plt.xlim(0)  # start x-axis from 0
plt.ylim(0)  # start y-axis from 0
plt.show()
```
